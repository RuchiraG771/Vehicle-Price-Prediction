```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
```

```python
data = pd.read_csv("../dataset/vehicle_data.csv")
data.head()
print(data.columns.tolist())
```

```python
data.info()
data.describe()

```

```python
data.isnull().sum()
data.dropna(inplace=True)
```

```python
data.columns
```

```python
plt.figure()
sns.histplot(data["price"], bins=30)
plt.title("Selling Price Distribution")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure()
sns.scatterplot(x="year", y="price", data=data)
plt.title("Vehicle Year vs Selling Price")
plt.show()
```

```python
plt.figure()
sns.boxplot(x="fuel", y="price", data=data)
plt.title("Fuel Type vs Selling Price")
plt.show()
plt.figure()
sns.boxplot(x="transmission", y="price", data=data)
plt.title("Transmission vs Selling Price")
plt.show()
```

```python
plt.figure(figsize=(10,6))

numeric_data = data.select_dtypes(include=[np.number])

sns.heatmap(numeric_data.corr(), annot=True, cmap="coolwarm")

plt.title("Feature Correlation Heatmap (Numeric Features Only)")
plt.show()

```

## 🔍 Key Insights from EDA

- Selling price **decreases as vehicle age increases**
- Cars with **lower km driven** have higher resale value
- **Automatic transmission** vehicles are generally priced higher
- **Diesel vehicles** show better resale value than petrol
- Engine power and mileage strongly influence price
